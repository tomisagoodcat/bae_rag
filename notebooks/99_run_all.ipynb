Thin wrapper around `run_pipeline.py`. Same logic as double-clicking `run_pipeline.bat`.

**Prerequisites:** Neo4j running, `DEEPSEEK_API_KEY`, papers in `data/markdown/forTest`, schema in `kg_build_pipeline/schema/`.

```python
import subprocess
import sys
from pathlib import Path

REPO = Path.cwd() if (Path.cwd() / "kg_build_pipeline").is_dir() else Path.cwd().parent
script = REPO / "kg_build_pipeline" / "run_pipeline.py"
config = REPO / "kg_build_pipeline" / "config.yaml"

cmd = [sys.executable, str(script), "--config", str(config), "--all"]
print(" ".join(cmd))
raise SystemExit(subprocess.call(cmd, cwd=str(REPO)))
```
